```python
import random
import time

distance = 0  # meters
speed = 0  # meters per second
fuel = 100.0  # percent
seconds = 0

print("🟢 Race Started! Monitoring live telemetry...")

while distance < 1000 and fuel > 0:
    seconds += 1

    # SIMULATION LOGIC: Let's randomly decide to Accelerate (2), Coast (1), or Brake (0)
    driver_action = random.choice([0, 1, 2])

    if driver_action == 2:  # Accelerate
        speed += 4
        fuel -= 1.5  # Heavy fuel burn
    elif driver_action == 1:  # Coast
        speed = max(0, speed - 1)  # Slow down slightly due to wind friction
        fuel -= 0.1  # Tiny fuel burn
    else:  # Brake
        speed = max(0, speed - 5)  # Quick slowdown
        fuel -= 0.0  # No fuel burn

    distance += speed

    # Print live telemetry stream
    print(
        f"Time: {seconds}s | Action: {driver_action} | Speed: {speed}m/s | Distance: {distance}m | Fuel: {fuel:.1f}%"
    )
    time.sleep(0.1)  # Slows down the screen output so you can read it live

print("\n🏁 Race Finished!")
print(f"Total Time: {seconds} seconds. Remaining Fuel: {fuel:.1f}%")
```

```python
import matplotlib.pyplot as plt
import time

# 1. INITIALIZE CAR STATE
distance = 0
speed = 0
fuel = 100.0
seconds = 0

# 2. CREATE EMPTY LOGS FOR THE GRAPH
time_data = []
speed_data = []
fuel_data = []

print("🟢 Race Started! Smart Strategy Engaged...")

while distance < 1000 and fuel > 0:
    seconds += 1
    
    # --- 🧠 THE PIT WALL LOGIC (OPTIMIZATION) ---
    dist_remaining = 1000 - distance
    
    # Strategy A: Panic Mode (Low fuel, far from finish)
    if fuel < 20 and dist_remaining > 300:
        target_speed = 12  
    # Strategy B: Sprint Mode (Near finish, plenty of fuel)
    elif dist_remaining < 200 and fuel > 15:
        target_speed = 35  
    # Strategy C: Efficient Cruise Control (Standard running)
    else:
        target_speed = 22  
        
    # Actuate the pedals based on the strategy
    if speed < target_speed:
        action = "Accelerate"
        speed += 4
        fuel -= 1.5
    elif speed > target_speed + 2:
        action = "Brake"
        speed = max(0, speed - 5)
        # Brakes don't burn fuel
    else:
        action = "Coast"
        speed = max(0, speed - 1)
        fuel -= 0.1
        
    distance += speed
    
    # --- 📊 LOGGING THE DATA (TELEMETRY) ---
    time_data.append(seconds)
    speed_data.append(speed)
    fuel_data.append(fuel)

print(f"🏁 Race Finished in {seconds} seconds!")
print(f"Final Distance: {distance}m | Remaining Fuel: {fuel:.1f}%")

# --- 📈 THE VISUALIZATION ---
plt.figure(figsize=(10, 5))

# Plot Speed (Blue line)
plt.plot(time_data, speed_data, label='Speed (m/s)', color='#1f77b4', linewidth=2.5)

# Plot Fuel (Red dashed line)
plt.plot(time_data, fuel_data, label='Fuel Level (%)', color='#d62728', linestyle='--', linewidth=2)

# Formatting the graph
plt.title("Smart Driver Telemetry Analysis", fontsize=14, fontweight='bold')
plt.xlabel("Race Time (Seconds)")
plt.ylabel("Magnitude")
plt.legend(loc="lower left")
plt.grid(True, linestyle=':', alpha=0.7)

# Show the dashboard
plt.show()
```
